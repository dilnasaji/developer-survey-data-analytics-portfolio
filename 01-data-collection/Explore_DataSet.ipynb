### 5. Estimate the average age

The Age column contains age ranges, not exact ages. To calculate a mean, we assign one representative number to each range. For example, the 18-24 group is represented by 21. These representative values are approximations, so the result is an estimated average, not the participant's exact average age. The open-ended groups ('Under 18'and '65 years or older') require especially strong assumptions.

First let's map age ranges to approximate numeric values


```python
age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

# Create a new numeric Age column
df['Age_numeric'] = df['Age'].map(age_mapping)
```

```python
## Write your code here
# Calculate the mean age
mean_age = df['Age_numeric'].mean()
print("mean age of the survey participants:", mean_age)
```

**Interpretation:** The saved output gives an estimated mean age of about 32.7 years using the representative values chosen above. This number depends on those choices and on which age labels matched the mapping. It should be described as an approximate summary of the age groups, not as an exact population average.

### 6. Count distinct countries

```python
## Write your code here
unique_countries = df["Country"].nunique()
print("Number of unique countries:", unique_countries)
```

**Interpretation:** The saved output reports 185 distinct country values. nunique() excludes missing values by default. This count reflects distinct labels in the Country column; it does not by itself verify whether labels are standardized or whether every value represents a sovereign country.
