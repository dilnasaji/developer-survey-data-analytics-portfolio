**Step 1: Count records for a technology**

The Key Skills field describes skills associated with a listing. Our first function demonstrates a case-insensitive exact-field match. For example, Python matches python, but it will not match Python, SQL or Python|SQL. This makes the count conservative for multi-skill listings.

Interpretation rule: The output is the number of records whose complete Key Skills field equals the requested label, not necessarily every job requiring that skill.

```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

def get_number_of_jobs_T(technology):
     
        response = requests.get(api_url)
        data = response.json()
        
        number_of_jobs=0
        
        for job in data:
            if technology.lower() == job.get("Key Skills", "").lower():
                    number_of_jobs = number_of_jobs + 1
            
        return technology, number_of_jobs
```

Calling the function for Python and checking if it works.


```python
get_number_of_jobs_T("Python")
```

#### Write a function to find number of jobs in US for a location of your choice


**Step 2: Count records for a location**

This function uses the same logic on the Location field. A count of 640 for Los Angeles means 640 dataset records have that exact location label in the saved notebook output; it does not mean there are 640 jobs currently open there.

```python
 def get_number_of_jobs_L(location):
    
        response = requests.get(api_url)
        data = response.json()
    
        number_of_jobs = 0
        
        for job in data:
            if location.lower() == job.get("Location", "").lower():
                   number_of_jobs = number_of_jobs + 1
                
        return location,number_of_jobs
```

```python
#your code goes here
get_number_of_jobs_L("Los Angeles")
```
