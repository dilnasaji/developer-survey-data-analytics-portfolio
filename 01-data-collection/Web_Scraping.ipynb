# **Web Scraping Programming Language Salary Data**

### Project overview

Salary comparisons often begin with information published on webpages rather than in ready-to-analyze spreadsheets. This guided exercise demonstrates how to collect a small table of programming languages and their listed average annual salaries from an IBM-hosted practice webpage, organize it with Python, and save it as a CSV file.

**Central question:** How can we turn a webpage table into a structured dataset that can be inspected or reused?

**What this project demonstrates:** HTTP requests, HTML parsing, table extraction, DataFrame creation, and CSV export.

**Scope and caution:** The figures are provided by the course's practice webpage. The notebook does not verify their source, methodology, date, geography, or currency beyond what appears on that page. They should not be presented as current market salary benchmarks.

Estimated time needed: **30 to 45** minutes


### Workflow at a glance
1. Identify the webpage containing the source table.
2. Download its HTML with requests.
3. Parse the HTML with BeautifulSoup.
4. Extract language names and salary text into a list of records.
5. Convert the records into a Pandas DataFrame and export to CSV.

## Extract information from the given web site
You will extract the data from the below web site: <br> 


##### Step 1: Identify the source
The URL points to a course-hosted HTML page. Unlike a CSV, an HTML page is designed for a browser, so we first need to retrieve and interpret its underlying markup.

```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**. It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.

Import the required libraries


##### Step 2: Choose the tools
requests retrieves webpage content; BeautifulSoup helps locate HTML elements such as tables and rows; and pandas turns the extracted records into a dataset.

```python
# Your code here
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

Download the webpage at the url


##### Step 3: Retrieve the webpage
The HTTP request downloads the page as text. This is the raw HTML, not yet a structured table. In a production scraper, we would also check the response status and handle failed requests.

```python
#your code goes here
data = requests.get(url).text
```

Create a soup object


##### Step 4: Parse HTML
Parsing makes it possible to find the table and its rows using HTML tags rather than manually copying information from the page.

```python
#your code goes here
soup = BeautifulSoup(data, "html.parser")
```

Scrape the `Language name` and `annual average salary`.


###### Step 5: Extract the two fields
The code locates the table, reads each row, and collects the programming language and salary into dictionaries. A DataFrame then organizes these records into two labeled columns.

**How to interpret the output:** Each row represents one language listed on the practice page; the salary is preserved as displayed text, including its currency symbol and separators. It is not yet a numeric column suitable for statistical calculations.

```python
#your code goes here
table = soup.find('table')

language_list = []

for row in table.find_all('tr'):
    cols = row.find_all('td')
    
    language_name = cols[1].getText()
    average_salary = cols[3].getText()
    
    language_list.append({"Language": language_name, "Average Salary": average_salary})
    
df_language = pd.DataFrame(language_list[1:])
print(df_language)
```

##### What the extracted table tells us
The saved notebook output shows 10 programming languages, including Python, Java, JavaScript, Swift, and SQL. In that snapshot, Swift has the highest listed salary ($130,801) and PHP has the lowest ($84,727). These are descriptions of the supplied practice table,not claims about present-day compensation.

A useful next step would be to remove $ and commas, convert salaries to numeric values, and create a comparison chart. That additional analysis is not part of the original completed exercise, so it is not represented here as work already performed.

Save the scrapped data into a file named *popular-languages.csv*


##### Step 6: Export for reuse
Writing the DataFrame to popular-languages.csv makes the collected information available to spreadsheet tools and future analysis. index=False prevents Pandas from adding an extra row-number column.

```python
# your code goes here
df_language.to_csv("popular-languages.csv", index=False)
print("File saved")
```

This exercise shows how to transform a human-readable webpage into a reusable dataset without manually transcribing its contents. The key technical sequence is request → parse → extract → structure → export.

**Skills demonstrated:** Python, Requests, BeautifulSoup, HTML table parsing, Pandas DataFrames, and CSV output.

**Limitations:** The script assumes the source page has the expected table structure and that the desired values are in specific columns. Changes to the HTML may break the extraction. The salary figures also lack enough context to support reliable comparisons across real-world job markets.

**Potential improvements:** Check HTTP responses, validate the table structure, handle missing cells, convert salary text to numeric values, and document the data source and collection date. These are proposed enhancements, not claims about the current implementation.
